```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import matplotlib.dates as mdates

# Force white plots regardless of PyCharm's theme
plt.style.use("default")

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "savefig.transparent": False,
    "text.color": "black",
    "axes.labelcolor": "black",
    "axes.titlecolor": "black",
    "axes.edgecolor": "black",
    "xtick.color": "black",
    "ytick.color": "black",
    "grid.color": "#b0b0b0",
})

LIGHTCURVE_COLOR = "#1f77b4"

# ============================================================
# 1. ENTER THE J0607-0834 DATA
# ============================================================

data = [
    ("1995-07-28", "U", 2.798),
    ("1996-06-05", "C", 2.384),
    ("1996-10-27", "U", 1.753),
    ("1997-01-11", "S", 2.435),
    ("1997-01-11", "X", 1.962),
    ("1997-05-07", "S", 3.371),
    ("1997-05-07", "X", 2.388),
    ("1998-10-30", "U", 1.389),
    ("2000-01-31", "S", 1.845),
    ("2000-01-31", "X", 1.272),
    ("2000-08-25", "U", 1.820),
    ("2001-03-15", "U", 1.542),
    ("2001-06-20", "U", 1.664),
    ("2001-08-25", "U", 1.495),
    ("2002-05-08", "S", 1.496),
    ("2002-05-08", "X", 1.335),
    ("2002-11-02", "U", 1.491),
    ("2003-02-05", "U", 1.483),
    ("2004-06-11", "U", 1.479),
    ("2005-03-05", "U", 1.437),
    ("2005-07-24", "U", 1.510),
    ("2005-09-16", "U", 1.361),
    ("2006-11-10", "U", 1.089),
    ("2007-06-10", "U", 1.024),
    ("2008-07-30", "U", 1.024),
    ("2009-05-02", "U", 1.327),
    ("2010-02-11", "U", 2.012),
    ("2010-08-23", "L", 1.102),
    ("2010-11-13", "U", 2.272),
    ("2011-01-15", "X", 1.686),
    ("2011-06-24", "U", 2.458),
    ("2012-03-27", "U", 2.844),
    ("2014-02-23", "C", 2.610),
    ("2014-02-23", "X", 4.180),
    ("2015-06-19", "Q", 1.585),
    ("2015-12-11", "C", 3.033),
    ("2015-12-11", "X", 3.176),
    ("2016-05-27", "C", 2.679),
    ("2016-05-27", "X", 2.798),
    ("2018-06-04", "C", 2.927),
    ("2018-06-04", "X", 2.015),
    ("2018-07-21", "S", 2.207),
    ("2018-07-21", "X", 2.504),
    ("2019-02-02", "C", 2.502),
    ("2019-02-02", "X", 2.119),
    ("2020-11-04", "C", 2.446),
    ("2020-11-04", "X", 1.847),
    ("2021-03-07", "S", 2.367),
    ("2021-03-07", "X", 2.842),
    ("2021-08-07", "U", 2.331),
    ("2021-09-30", "C", 2.841),
    ("2021-09-30", "X", 2.709),
    ("2021-12-12", "S", 2.280),
    ("2021-12-12", "X", 2.458),
    ("2022-02-24", "U", 2.397),
    ("2022-07-29", "C", 2.633),
    ("2022-07-29", "X", 1.925),
    ("2022-08-09", "U", 2.267),
    ("2022-10-23", "S", 2.318),
    ("2022-10-23", "X", 1.908),
    ("2023-05-03", "U", 2.220),
    ("2023-11-14", "U", 2.243),
    ("2024-04-06", "S", 2.427),
    ("2024-04-06", "X", 2.216),
    ("2024-05-17", "U", 2.134),
    ("2024-07-28", "K", 1.282),
    ("2024-11-26", "U", 2.244),
    ("2025-03-16", "U", 2.306),
    ("2025-10-26", "U", 2.026),
    ("2026-06-28", "C", 2.159),
    ("2026-06-28", "X", 2.371),
    ("2026-07-21", "C", 2.606),
    ("2026-07-21", "X", 2.326),
]


# ============================================================
# 2. CREATE DATAFRAME
# ============================================================

df = pd.DataFrame(
    data,
    columns=["date", "band", "flux"]
)


# Convert date into datetime
df["date"] = pd.to_datetime(df["date"])


# ============================================================
# 3. CONVERT DATE TO MJD
# ============================================================

# Unix epoch:
# 1970-01-01 corresponds to MJD = 40587

df["mjd"] = (
    df["date"].astype("int64")
    / 1e9
    / 86400
    + 40587
)


# ============================================================
# 4. CREATE OUTPUT FOLDER
# ============================================================

os.makedirs(
    "J0607_0834_band_lightcurves",
    exist_ok=True
)


# ============================================================
# 5. PLOT EACH BAND SEPARATELY
# ============================================================

bands = df["band"].unique()

for band in bands:

    # Select only this band
    band_data = df[
        df["band"] == band
    ].copy()

    # Sort chronologically
    band_data = band_data.sort_values("date")


    # --------------------------------------------------------
    # Create figure
    # --------------------------------------------------------

        # --------------------------------------------------------
    # Create figure with a white background
    # --------------------------------------------------------

    fig, ax = plt.subplots(
        figsize=(10, 5),
        facecolor="white"
    )

    fig.patch.set_facecolor("white")
    ax.set_facecolor("white")

    # Blue light-curve points
    ax.scatter(
        band_data["date"],
        band_data["flux"],
        s=35,
        color=LIGHTCURVE_COLOR,
        edgecolors=LIGHTCURVE_COLOR
    )

    event_date = pd.Timestamp("2023-02-13")

    # Black event-date line
    ax.axvline(
        event_date,
        color="black",
        linestyle="--",
        linewidth=1.5,
        label="13 Feb 2023"
    )

    # White legend with black text
    legend = ax.legend(
        facecolor="white",
        edgecolor="grey",
        framealpha=1
    )

    for legend_text in legend.get_texts():
        legend_text.set_color("black")

    ax.xaxis.set_major_locator(
        mdates.AutoDateLocator()
    )

    ax.xaxis.set_major_formatter(
        mdates.DateFormatter("%Y")
    )

    # --------------------------------------------------------
    # Labels
    # --------------------------------------------------------

    ax.set_xlabel(
        "Date",
        color="black"
    )

    ax.set_ylabel(
        "Total Flux Density (Jy)",
        color="black"
    )

    ax.set_title(
        f"J0607-0834 - {band} Band Lightcurve "
        f"(N = {len(band_data)})",
        color="black"
    )

    ax.tick_params(
        axis="both",
        colors="black"
    )

    ax.grid(
        True,
        color="#b0b0b0",
        alpha=0.3,
        linestyle="--"
    )

    plt.setp(
        ax.get_xticklabels(),
        rotation=45,
        ha="right",
        color="black"
    )

    fig.tight_layout()

    # --------------------------------------------------------
    # Save each band separately
    # --------------------------------------------------------

    filename = (
        "J0607_0834_band_lightcurves/"
        f"J0607-0834_{band}_band_lightcurve.png"
    )

    fig.savefig(
        filename,
        dpi=300,
        bbox_inches="tight",
        facecolor="white",
        edgecolor="white",
        transparent=False
    )

    plt.close(fig)


    print(
        f"{band} band: "
        f"{len(band_data)} points saved"
    )


print("\nDone.")
```
