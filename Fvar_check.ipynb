```python
import pandas as pd
from astropy.time import Time
import matplotlib.pyplot as plt
import os
import numpy as np
from astropy.stats import sigma_clip
```

```python
df = pd.read_csv("km3netevent_lightcurve_tiles.csv")

master_id = 1007034027074

# Select this source
lc = df.loc[
    pd.to_numeric(df["MasterID"], errors="coerce") == master_id,
    ["Mag", "Magerr"]
].copy()

# Remove invalid values
lc["Mag"] = pd.to_numeric(lc["Mag"], errors="coerce")
lc["Magerr"] = pd.to_numeric(lc["Magerr"], errors="coerce")

lc = lc.dropna()
lc = lc[
    np.isfinite(lc["Mag"])
    & np.isfinite(lc["Magerr"])
    & (lc["Magerr"] > 0)
].copy()

# 3-sigma clipping
clipped = sigma_clip(
    lc["Mag"].to_numpy(),
    sigma=3,
    maxiters=None,
    cenfunc="median",
    stdfunc="mad_std"
)

lc = lc.loc[~np.ma.getmaskarray(clipped)].copy()

# Convert magnitude to relative flux
lc["flux"] = 10**(-0.4 * lc["Mag"])

# Propagate magnitude error to flux error
lc["flux_err"] = (
    0.4 * np.log(10) * lc["flux"] * lc["Magerr"]
)

x = lc["flux"].to_numpy()
sigma_err = lc["flux_err"].to_numpy()

N = len(x)

# Quantities in equation (1)
mean_x = np.mean(x)
S2 = np.var(x, ddof=1)
mean_sigma_err2 = np.mean(sigma_err**2)

# Equation (1)
Fvar = np.sqrt(
    (S2 - mean_sigma_err2) / mean_x**2
)

# Equation (3)
err_sigma_NXS2 = np.sqrt(
    (
        np.sqrt(2 / N)
        * mean_sigma_err2
        / mean_x**2
    )**2
    +
    (
        np.sqrt(mean_sigma_err2 / N)
        * 2 * Fvar
        / mean_x
    )**2
)

# Equation (2)
Fvar_err = np.sqrt(
    Fvar**2 + err_sigma_NXS2
) - Fvar

print("Fvar =", Fvar)
print("Fvar error =", Fvar_err)
```
